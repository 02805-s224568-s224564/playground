```python
# Example figure: keep, rename or delete. Every figure in the post must come from
# save_figure(): it writes assets/figures/week__WEEK__-<slug>.png and prints the
# include line to paste into the post. Alt text is required.
deg_in = pd.Series(dict(G.in_degree()), name="in")
deg_out = pd.Series(dict(G.out_degree()), name="out")

fig, ax = plt.subplots()
ax.scatter(deg_out, deg_in, s=18, alpha=0.7)
ax.set(xlabel="Out-degree (links from the article)", ylabel="In-degree (links to the article)")
for node in deg_in.nlargest(3).index:
    ax.annotate(G.nodes[node]["name"], (deg_out[node], deg_in[node]),
                fontsize=8, xytext=(4, 2), textcoords="offset points")

print(save_figure(fig, "example-in-vs-out-degree", week=WEEK,
                  alt="Scatter of in-degree against out-degree for the 303 characters",
                  caption="Replace this caption, or delete the figure."))
```
